# Game Runner in a Lesson

A Game Runner connects JavaScript game code to the GameEngine framework. The runner supplies Start, Pause, Stop, Fullscreen, Save, and Reset controls while the game code supplies the level classes.

## Assignment 1: Basic Custom Level

Run the game and use WASD or the arrow keys to move Chill Guy around the desert-style background. Try changing the starting position or movement speed in the editor.

```python
%%js

// GAME_RUNNER: Run the game and use WASD or arrow keys to move Chill Guy around the level.

import GameControl from '/assets/js/GameEnginev1/essentials/GameControl.js';
import GameEnvBackground from '/assets/js/GameEnginev1/essentials/GameEnvBackground.js';
import Player from '/assets/js/GameEnginev1/essentials/Player.js';

class CustomLevel {
  constructor(gameEnv) {
    const path = gameEnv.path;
    const bgData = {
      name: 'custom_bg',
      src: path + '/images/gamebuilder/bg/clouds.jpg',
      pixels: { height: 720, width: 1280 }
    };
    const playerData = {
      id: 'Hero',
      src: path + '/images/gamify/chillguy.png',
      SCALE_FACTOR: 5,
      STEP_FACTOR: 1000,
      ANIMATION_RATE: 50,
      INIT_POSITION: { x: 100, y: 300 },
      pixels: { height: 512, width: 384 },
      orientation: { rows: 4, columns: 3 },
      down: { row: 0, start: 0, columns: 3 },
      downRight: { row: 1, start: 0, columns: 3 },
      downLeft: { row: 2, start: 0, columns: 3 },
      right: { row: 1, start: 0, columns: 3 },
      left: { row: 2, start: 0, columns: 3 },
      up: { row: 3, start: 0, columns: 3 },
      upRight: { row: 1, start: 0, columns: 3 },
      upLeft: { row: 2, start: 0, columns: 3 },
      hitbox: { widthPercentage: 0.45, heightPercentage: 0.2 },
      keypress: { up: 87, left: 65, down: 83, right: 68 }
    };

    this.classes = [
      { class: GameEnvBackground, data: bgData },
      { class: Player, data: playerData }
    ];
  }
}

export const gameLevelClasses = [CustomLevel];
export { GameControl };
```

## Assignment 2: Combine Game Levels

Run the game, open the level selector, and switch between the water and fish levels. This shows how one runner can provide multiple playable experiences.

```python
%%js

// GAME_RUNNER: Switch between the water and fish levels using the level selector.

import GameControl from '/assets/js/GameEnginev1/essentials/GameControl.js';
import GameLevelWater from '/assets/js/GameEnginev1/GameLevelWater.js';
import GameLevelParallaxFish from '/assets/js/GameEnginev1/GameLevelParallaxFish.js';

export const gameLevelClasses = [GameLevelWater, GameLevelParallaxFish];
export { GameControl };
```

## Game Runner Checklist

- Import `GameControl` and the level classes you need.
- Define a level class with a constructor and a `this.classes` array.
- Export `GameControl` and `gameLevelClasses`.
- Give each runner a different challenge marker and generated runner ID.
- Use Start, Pause, Stop, Reset, and the level selector while testing.
