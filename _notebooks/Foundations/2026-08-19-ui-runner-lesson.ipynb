canvas.width = 400;
canvas.height = 400;
outputElement.appendChild(canvas);

## Assignment 2: Create an Interactive Button

Create a button with JavaScript and make it change color and text each time it is clicked. This assignment practices DOM creation, styling, and event listeners.

{% capture challenge2 %}
Create an interactive button that changes color when clicked. Use DOM manipulation to create and style the button.
{% endcapture %}

{% capture code2 %}
outputElement.innerHTML = '';

const button = document.createElement('button');
button.textContent = 'Click Me!';
button.style.padding = '20px 40px';
button.style.fontSize = '18px';
button.style.border = 'none';
button.style.borderRadius = '8px';
button.style.cursor = 'pointer';
button.style.backgroundColor = '#3b82f6';
button.style.color = 'white';

let isBlue = true;
button.addEventListener('click', () => {
    if (isBlue) {
        button.style.backgroundColor = '#ef4444';
        button.textContent = 'Red!';
    } else {
        button.style.backgroundColor = '#3b82f6';
        button.textContent = 'Blue!';
    }
    isBlue = !isBlue;
});

outputElement.appendChild(button);
{% endcapture %}

## Assignment 3: Build a Simple Level

Create a 400 by 400 canvas with a blue player square. Use the arrow keys or WASD to move the player while the animation loop redraws the level. This is a small game-style UI Runner assignment.

{% capture challenge3 %}
Use the canvas to create a simple level. Move the blue player with the arrow keys or WASD.
{% endcapture %}

{% capture code3 %}
outputElement.innerHTML = '';

const canvas = document.createElement('canvas');
canvas.width = 400;
canvas.height = 400;
canvas.style.border = '2px solid #3b82f6';
outputElement.appendChild(canvas);

const ctx = canvas.getContext('2d');
const keys = {};
let playerX = 200;
let playerY = 200;
const playerSize = 30;

const handleKeyDown = (event) => {
    keys[event.key.toLowerCase()] = true;
};
const handleKeyUp = (event) => {
    keys[event.key.toLowerCase()] = false;
};
document.addEventListener('keydown', handleKeyDown);
document.addEventListener('keyup', handleKeyUp);

function animate() {
    if (keys.arrowleft || keys.a) playerX -= 3;
    if (keys.arrowright || keys.d) playerX += 3;
    if (keys.arrowup || keys.w) playerY -= 3;
    if (keys.arrowdown || keys.s) playerY += 3;

    playerX = Math.max(playerSize / 2, Math.min(canvas.width - playerSize / 2, playerX));
    playerY = Math.max(playerSize / 2, Math.min(canvas.height - playerSize / 2, playerY));

    ctx.fillStyle = '#111827';
    ctx.fillRect(0, 0, canvas.width, canvas.height);
    ctx.fillStyle = '#3b82f6';
    ctx.fillRect(playerX - playerSize / 2, playerY - playerSize / 2, playerSize, playerSize);
    requestAnimationFrame(animate);
}

animate();
{% endcapture %}

## Short Summary

I learned that a UI Runner combines a JavaScript editor with an `outputElement` for live DOM and canvas output. I practiced drawing with Canvas, creating interactive buttons, handling keyboard input, and using animation loops. I also learned that unique runner IDs and clearing the output keep multiple interactive activities independent.

## My Unicorn Improvement

My unicorn improvement would be a live preview guide beside each assignment that highlights the canvas size, coordinates, and next step while students experiment. The lesson is interactive, but this guide would make the first attempt feel more welcoming and help students understand why each change affects the output.
