# Browser Code Runner: `indexOf` and `substring`

This activity practices the same string methods used in Java, but the runnable version uses JavaScript so it works directly in the browser without a backend. Run the code, inspect each result, and then change the test data.

## Assignment

Complete and test a function that removes the first occurrence of a target string. Pay attention to the case where the target is missing.

```python
%%js

// CODE_RUNNER: Methods and Control Structures - practice indexOf and substring in the browser

function shortenMessage(message, target) {
    console.log('Original Message: ' + message);
    console.log('Substring to remove: ' + target);

    const position = message.indexOf(target);
    if (position === -1) {
        console.log('No action, substring not found: ' + target);
        return message;
    }

    const length = target.length;
    const newFront = message.substring(0, position);
    const newBack = message.substring(position + length);

    console.log('Front: ' + newFront);
    console.log('Back: ' + newBack);
    return newFront + newBack;
}

const messageData = [
    ['Hello, OpenCS!', ','],
    ['Java code code coding!', 'Java'],
    ['Java programming is fun.', '.'],
    ['No substring here', 'exception']
];

for (const data of messageData) {
    const result = shortenMessage(data[0], data[1]);
    console.log('Shortened Message: ' + result);
    console.log('-----');
}
```
